```python
import pandas as pd 
import requests
from io import BytesIO
import streamlit as st
```

```python
@st.cache_data
def load_and_clean_data():
    url = "https://github.com/Abdourahmane2/donne_etudiant/raw/main/base_finale.csv"
    response = requests.get(url)
    if response.status_code != 200:
        st.error("❌ Échec du téléchargement.")
        st.stop()

    data = pd.read_csv(BytesIO(response.content), low_memory=False)
    return data
```

```python
#afficher le type de chaque colonne
data['moyenne_annuelle'] = pd.to_numeric(data['moyenne_annuelle'], errors='coerce')
data['moyenne_annuelle'].dtype
#remplacer les valeurs manquantes par la moyenne des etudiants similares 
data['moyenne_annuelle'] = data.groupby(['resultat', 'mention_bacc'])['moyenne_annuelle'].transform(lambda x: x.fillna(x.mean()))
```

```python
#supprimer la colonne mention car % de valeurs trop importante remplacer par la mode peut creer un desiquilibre  
data = data.drop(columns=['mention'])
```

```python
#regrouper les series bacc en 3 categories 
import pandas as pd
def serie_bac_type(serie):
    s = str(serie).upper().strip()
    if any(x in s for x in ['S', 'SCI', 'D', 'E', 'F6', 'F4', 'F5', 'S2', 'S3', 'S4', 'S5', 'SM', 'A2', 'A3', 'S1A', 'S2A']):
        return 'Scientifique'
    elif any(x in s for x in ['L', 'A', 'AR', 'L1', 'L2', 'L\'', 'L1B', 'L1A', 'LA', 'SNA', 'SB']):
        return 'Littéraire'
    elif any(x in s for x in ['G', 'T', 'STG', 'S1', 'S2', 'S3', 'S4', 'S5', 'T1', 'F7', 'CAEP', 'STIDD', 'STEG']):
        return 'Tertiaire'
    else:
        return 'Autre'

data['categorie_bac'] = data['serie_bacc'].apply(serie_bac_type)




```

```python
data['mention_bacc'].unique()
```

```python
#gerer les mentions bacc
def clean_mention(mention):
    mention = str(mention).upper().strip()
    # Standardiser les mentions valides
    if mention == 'AB' or mention == 'Ab':
        return 'Assez Bien'
    elif mention == 'TB' or mention == 'TH':
        return 'Très Bien'
    elif mention == 'BI':
        return 'Bien'
    elif mention == 'A':
        return 'Assez Bien'
    elif mention == 'PA':
        return 'Passable'
    elif mention in ['KA', 'GU', 'MB', 'LO', '..', 'BA', 'LV', 'RU', 'HO' , '50']:
        return 'non renseigne'  


# Appliquer la fonction de nettoyage à la colonne 'mention_bacc'
data['mention_bacc'] = data['mention_bacc'].apply(clean_mention)

#remplaceer les mentions non renseignées par la mention la plus fréquente
most_frequent_mention = data['mention_bacc'].mode()[0]
data['mention_bacc'] = data['mention_bacc'].replace('non renseigne', most_frequent_mention)
data['mention_bacc'].fillna(most_frequent_mention , inplace=True)

```

```python
#annee de bac 
data['annee_bacc'].fillna(data['annee_bacc'].mode()[0] , inplace=True) #remplacer par la valeur la plus fréquente
```

```python
#supprimer les lignes avec des valeurs manquantes dans 'annee_bacc'
data = data.dropna(subset=['moyenne_annuelle', 'credit', 'session' , 'resultat'])
```

```python

# Fonction pour filtrer les étudiants en fonction des critères
def filtrer_etudiants(df, age_min, age_max, annee_bac , serie_bacc, nationalite):
    # Convertir la colonne 'date_de_naissance' en format datetime
    df['date_de_naissance'] = pd.to_datetime(df['date_de_naissance'], errors='coerce')

    # Calculer l'âge à partir de la date de naissance
    df['age'] = df['annee_bacc'] - df['date_de_naissance'].dt.year
    
    # Filtrer les étudiants en fonction des critères fournis
    df_filtre = df[
        (df['age'] >= age_min) & (df['age'] <= age_max) & 
        (df['serie_bacc'] == serie_bacc) & 
        (df['nationalite'] == nationalite) &
        (df['annee_bacc'] == annee_bac -1)  #prendre que les etudiands inscrites l'annee d'avant
    ]
    return df_filtre

# Fonction pour générer les statistiques par mention de bac et département
def statistiques_par_mention_bac(df_filtre):
    if df_filtre.empty:
        return "Aucun étudiant correspondant à ces critères."

    # Dictionnaire pour stocker les résultats
    stats_list = []
    
    for mention in df_filtre['mention_bacc'].unique():
        # Filtrer par mention
        df_mention = df_filtre[df_filtre['mention_bacc'] == mention]
        
        # Calcul des statistiques par département
        for departement in df_mention['departement_formation'].unique():
            # Filtrer par département
            df_departement = df_mention[df_mention['departement_formation'] == departement]
            
            # Si aucun étudiant n'est orienté dans ce département, on le saute
            if df_departement.empty:
                continue
            
            # Calcul du taux de réussite (résultat == "Passe")
            taux_reussite = (df_departement['resultat'] == 1).mean() * 100
            
            # Calcul du taux de redoublement (résultat == "Redouble")
            taux_redoublement = (df_departement['resultat'] ==0).mean() * 100
            
            # Calcul de la moyenne obtenue dans ce département
            moyenne_departement = df_departement['moyenne_annuelle'].mean()
            
            # Nombre d'étudiants orientés dans ce département
            nombre_etudiants = len(df_departement)
            
            # Ajouter les résultats dans la liste
            stats_list.append({
                'Mention de Bac': mention,
                'Département': departement,
                'Taux de réussite': taux_reussite,
                'Taux de redoublement': taux_redoublement,
                'Moyenne obtenue': moyenne_departement,
                'Nombre d\'étudiants orientés': nombre_etudiants
            })

    # Convertir les résultats en DataFrame
    stats_df = pd.DataFrame(stats_list)
    
    return stats_df

# Exemple pour l'étudiant
age_min = 18
age_max = 20
serie_bacc = 'L2'  # Série scientifique
nationalite = 'Sénégalaise'
annee_bac = 2023  # Année de baccalauréat

# Filtrer les étudiants selon les critères
df_similaires = filtrer_etudiants(data, age_min, age_max, annee_bac , serie_bacc, nationalite)

# Calculer les statistiques des étudiants filtrés par mention de bac
stats_similaires_mention = statistiques_par_mention_bac(df_similaires)

# Affichage des résultats sous forme de DataFrame 
print("voici quelques statistiques  des etudiants similaires orientes l'année d'avant :") 
stats_similaires_mention
```
